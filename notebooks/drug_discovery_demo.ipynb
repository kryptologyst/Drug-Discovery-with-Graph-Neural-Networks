# Drug Discovery with Graph Neural Networks

This notebook demonstrates how to use Graph Neural Networks for molecular property prediction in drug discovery.

## Overview

We'll train and evaluate different GNN architectures on the ESOL dataset for aqueous solubility prediction:
- **GCN**: Graph Convolutional Network
- **GIN**: Graph Isomorphism Network
- **GraphSAGE**: Graph Sample and Aggregate
- **MPNN**: Message Passing Neural Network
